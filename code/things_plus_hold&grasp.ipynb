```python
import pandas as pd
import numpy as np
from pathlib import Path

base_dir = Path(
    r"D:\NTU_PSY\Documents\Patric_Asen_BrainHackProject"
)

df = pd.read_csv(base_dir / "things_plus/02_object-level/_property-ratings.tsv", sep='\t')

print(df.shape)
print(df.columns)
df.head()
```

```python
df['grasp'] = df['property_grasp_mean']
df['hold'] = df['property_hold_mean']

print(df['grasp'].describe())
print(df['hold'].describe())
```

```python
df_grasp = df[['uniqueID', 'property_grasp_mean']].copy()
df_hold = df[['uniqueID', 'property_hold_mean']].copy()

df_grasp.rename(columns={'property_grasp_mean': 'grasp_rating'}, inplace=True)
df_hold.rename(columns={'property_hold_mean': 'hold_rating'}, inplace=True)

print(df_grasp.head())
print(df_hold.head())

df_grasp.to_csv(base_dir / 'grasp_ratings.csv', index=False)
df_hold.to_csv(base_dir / 'hold_ratings.csv', index=False)
```
