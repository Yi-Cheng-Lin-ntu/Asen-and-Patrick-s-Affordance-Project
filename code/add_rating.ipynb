```python
import pandas as pd
import numpy as np
from pathlib import Path

base_dir = Path(
    r"D:\NTU_PSY\Documents\Patric_Asen_BrainHackProject"
)

hold_rating = pd.read_csv(base_dir / "hold_ratings.csv")
grasp_rating = pd.read_csv(base_dir / "grasp_ratings.csv")

hold_rating = hold_rating.rename(columns={"uniqueID": "concept"})
grasp_rating = grasp_rating.rename(columns={"uniqueID": "concept"})

print(hold_rating.head())
print(grasp_rating.head())
```

```python
sub_list = ["sub-01", "sub-02", "sub-03"]

for sub in sub_list:
    print(f"\n===== Processing {sub} =====")

    sub_cond = pd.read_csv(base_dir / f"subject_fMRI_nii/{sub}_condition_clean.csv")

    print(set(sub_cond["concept"]) - set(grasp_rating["concept"]))
    print(set(sub_cond["concept"]) - set(hold_rating["concept"]))

    sub_cond_rating = sub_cond.merge(
    grasp_rating[["concept", "grasp_rating"]],
    on="concept",
    how="left")

    sub_cond_rating = sub_cond_rating.merge(
        hold_rating[["concept", "hold_rating"]],
        on="concept",
        how="left"
    )

    print(sub_cond_rating[["grasp_rating", "hold_rating"]].isna().sum())
    print(sub_cond_rating.head())

    sub_cond_rating.to_csv(base_dir / f"subject_fMRI_nii/{sub}_condition_with_ratings.csv", index=False)
```
