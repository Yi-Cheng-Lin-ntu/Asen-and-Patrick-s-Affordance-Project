```python
from pathlib import Path
import gzip
import shutil

root_dir = Path(
    r"D:\NTU_PSY\Documents\Patric_Asen_BrainHackProject\subject_fMRI_nii"
)

subjects = sorted(root_dir.glob("sub-*"))

for subject_dir in subjects:

    print(f"\nProcessing {subject_dir.name}")

    gz_files = list(subject_dir.rglob("*.nii.gz"))

    print("found:", len(gz_files))

    for gz_path in gz_files:

        session_dir = next(
            (p for p in gz_path.parents if p.name.startswith("ses-")),
            None
        )

        if session_dir is None:
            continue

        unzip_dir = session_dir / "nii_unzipped"
        unzip_dir.mkdir(exist_ok=True)

        out_path = unzip_dir / gz_path.stem

        if out_path.exists():
            continue

        with gzip.open(gz_path, "rb") as f_in:
            with open(out_path, "wb") as f_out:
                shutil.copyfileobj(f_in, f_out)

print("\nAll subjects done")
```
