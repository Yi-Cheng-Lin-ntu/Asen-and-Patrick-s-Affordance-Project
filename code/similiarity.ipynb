```python
import numpy as np
import pandas as pd
```

```python
import scipy.io as sio

mat_RDM_1 = sio.loadmat("THINGS-behavior/data/spose_similarity.mat")
mat_RDM_2 = sio.loadmat("THINGS-behavior/data/embedding49/spose_similarity49.mat")
print(mat_RDM_1.keys())
print(mat_RDM_2.keys())
```

```python
A = mat_RDM_1["spose_sim"]
B = mat_RDM_2["spose_sim"]

print(A.shape)
print(B.shape)

print(np.array_equal(A, B))
print(np.allclose(A, B, atol=1e-6))
```

```python
diff = np.abs(A - B)

print("max diff:", np.max(diff))
print("mean diff:", np.mean(diff))
```

```python
from scipy.stats import pearsonr

r, p = pearsonr(A.flatten(), B.flatten())
print(r)
```
